```python
import networkx as nx
import random
import matplotlib.pyplot as plt

grid_size = int(input('Enter grid size (5 for default): ') or 5)
edges_to_remove = int(input('Edges to remove: ') or 0)
total_vertices = grid_size * grid_size
print(f"Total vertices: {total_vertices}, edges to remove: {edges_to_remove}")
```

```python
G_full = nx.grid_2d_graph(grid_size, grid_size)
full_edges_count = G_full.number_of_edges()
print(f"Graph grid size: {grid_size} x {grid_size} ({total_vertices} vertices) \n"
      f"Starting edges count: {full_edges_count}")

pos = {}
for node in G_full.nodes():
    x = node[1]
    y = grid_size - 1 - node[0]
    pos[node] = (x, y)
```

```python
G_road = G_full.copy()
all_edges = list(G_road.edges())
random.shuffle(all_edges)
removed_count = 0
edges_to_keep_removed = []
print(f"Trying to remove {edges_to_remove} edges...")
```

```python
for edge in all_edges:
    if removed_count >= edges_to_remove:
        break
    G_road.remove_edge(*edge)

    if nx.is_connected(G_road):
        removed_count += 1
        edges_to_keep_removed.append(edge)
    else:
        G_road.add_edge(*edge)

print(f"Deleted edges: {removed_count} \n"
      f"Edges in final graph: {G_road.number_of_edges()}")
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title("Starting grid graph")
nx.draw(G_full, pos, node_color="lightgray", edgecolors="black", node_size=100)

plt.subplot(1,2,2)
plt.title(f"Road graph ({removed_count} edges removed)")
nx.draw(G_road, pos, node_color="white", edgecolors="black", node_size=100)
plt.show()
```
